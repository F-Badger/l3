# Recommender systems 

Recommender systems belong to a class of information filtering systems.  
They consist of algorithms/tools/techniques that:

- Take a user's preferences or behavioural data as input
- Predict a user's preference for, or likely interaction with, an item
- Output suggestions/recommendations about what a user may like

They combine:

- machine learning/data mining
- business and marketing
- human-computer interaction (HCI)

## Predictions and recommendations

Predictions are an important part of the process of a recommender system in producing recommendations.

Predictions:

- help to estimate the likely user preference or interaction with an item
- are often numerical values, e.g. a predicted rating or probability of clicking

Recommendations:

- are a selection of items from a larger set
  - e.g., the N items that the user is predicted to like or interact with the most
- may or may not be ordered (ranked)

# Content-based filtering

Content-based filtering (CBF):

- analyses the features of items/documents that the user has previously rated, liked, or interacted with
- builds a model/profile of the user's interests based on these features
- uses this profile to recommend items with features similar to those of items the user has liked in the past

## Benefits of CBF

The benefits of CBF include:

- user independence
  - does not rely on data about other users' preferences
- transparency
  - recommendations can often be explained by identifying the item features that match the user's interests
- easy to recommend new items
  - recommendations are based on item features rather than historical ratings, so new items can be recommended as long as their features are available
 
## Drawback of CBF

The drawbacks of CBF include:

- limited content analysis
  - the number and type of features available for each item may be limited
  - domain knowledge may be required to identify useful features, depending on the domain
- over-specialisation
  - tends to recommend items similar to those the user has liked in the past
  - has no inherent mechanism for discovering unexpected items or introducing users to different types of content
- difficult to provide recommendations for new users
  - recommendations may be unreliable because there is insufficient information about the user's preferences
  - this is known as the cold-start problem
- user profile updating
  - user interests may change over time, so the profile may need to be updated using new feedback
  - older ratings or interactions can be given less weight to account for changing preferences

## CBF architecture 

![CBF architecture](CBFarchitecture.png)

The architecture of a CBF system is generally as follows:

- information about items is fed into a content analyser
- the content analyser:
  - represents an item's content in a structured form
  - performs feature extraction (transforming the original information into a suitable feature representation)
  - items can be represented as:
    - structured data - a fixed set of attributes, with defined values or value types
    - unstructured data - e.g., text, posts, and opinions, from which useful features must be extracted
- information about items the user has previously interacted with is fed into a profile learner
- the profile learner:
  - collects and generalises information about the user's preferences
  - constructs the user profile
  - user profiles can be a vector model
    - represents the user's interests by aggregating the features of preferred items into a preference vector
    - recommendations are made by comparing this vector with the feature vectors of candidate items
    - e.g., a film preference vector of [0.8 action, 0.2 romance] indicates a stronger preference for action films than romance films
  - or, they can be a learned model
    - trains a model using the user's historical interactions to predict the probability of the user liking an item or the rating they would give it, based on its features
    - recommendations are made by scoring candidate items using the model and selecting those with the highest predicted scores
    - e.g., a model might predict a 90% probability that the user will like a particular action film
- the user profile and represented items are fed into a filtering component
- the filtering component:
  - generates a list of item recommendations by:
    - selecting eligible candidate items
    - scoring each candidate using the user's profile/model
    - sorting candidates by score and returning the top K items
    - excluding items the user has already consumed, where appropriate
  - collects feedback on recommendations, which can be used to update the user profile

## Content analysis

To perform content analysis, an item description is required. This description can consist of:

- structured data (attributes)
  - content analysis can be performed directly on these attributes
  - e.g., a film's genre, director, release year and language
- unstructured data
  - can be analysed using a keyword-based approach
    - extracts features based on keywords or terms in the content
    - may use predefined rules, statistical methods or machine learning
    - limited understanding of meaning; struggles with polysemy (multiple meanings of a word) and synonymy (different words with similar meanings)
  - can also be analysed using semantic analysis
    - uses knowledge bases, such as:
      - lexicons - collections of words and information about their meanings, synonyms and relationships
      - ontologies - formal representations of concepts within a domain and the relationships between them
    - aims to interpret the meaning of content in relation to the user's information needs
    - can better account for relationships between words and concepts

### Keyword-based approach pipeline

The pipeline for the keyword-based approach consists of:

- pre-processing
   - tokenisation - splits text into tokens (e.g., words, sentences or characters)
   - cleaning and normalisation - stopword removal, stemming, lemmatisation, lowercasing and spell-checking
- feature selection
  - calculates a metric for each feature and ranks features by their values
  - selects the most informative features, potentially removing irrelevant or redundant ones
- vector construction
  - combines the selected features into a shared feature space
  - represents each document/item as a vector of equal length, with each position corresponding to a feature
  - duplicate occurrences of a term do not create new features, although their frequency may be recorded
- weighting scheme
  - assigns a weight to each feature to represent its importance within a document/item
  - e.g., TF-IDF gives higher weight to terms that are frequent in a document but uncommon across the document collection

#### The item vector 

After vector construction, all items are represented by vectors of equal length, using the same terms/features in the same order.

A vector of term weights is a list of numbers representing the strength or importance of each term/feature for a particular item.

- the vector exists in an n-dimensional space, where n is the number of features in the dictionary
- a corpus is the complete collection of documents/items being analysed
- the dictionary is the set of selected terms/features extracted from the corpus; each term/feature corresponds to one dimension
  - e.g., for movies, features could include genres, actors and directors
- each weight represents the degree of association between an item and a term/feature, according to the weighting scheme used
  - e.g., a high weight for comedy indicates a strong association between a movie and the comedy feature
