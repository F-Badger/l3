# Nature signals

A natural signal is data that originates from the real world.

Examples include:

- Images: Pixel values arranged in two-dimensional space.
- Audio: Sound waves represented as sequences of values over time.
- Video: Sequences of images changing over time.
- Language: Words arranged in sequences with grammatical and semantic relationships.

These signals tend to exhibit three important properties:

- Repetitions in space and time
  - Natural signals often contain patterns that repeat across different spatial locations or points in time.
  - e.g. The same edge or texture can appear in different locations within an image.
- Various symmetries
  - A symmetry is a transformation of data that preserves some important property or structure
  - e.g. An object can move to a different position in an image while remaining the same object.
- Signals are hierarchical
  - Natural signals are often organised into multiple levels of increasingly complex structures
  - e.g. For a face:
    - Low-level features: Edges, lines and basic colours.
    - Intermediate features: Textures, shapes and combinations of edges.
    - High-level features: Eyes, noses and mouths.
    - Complex concepts: A complete face.

We can design neural network architectures to exploit the natural structure of the data:

- Rather than learning everything from scratch, these architectural choices introduce useful assumptions about the data.
- These assumptions are called inductive biases.
- Inductive biases help neural networks learn more efficiently by exploiting patterns that are already expected to exist in the data.

# Convolutional neural networks

A convolution is an operation that applies a small matrix of learnable weights (called a kernel or filter) to a larger input.

The kernel slides across the input, performing calculations at each position to produce an output value.

At each position:

1. Take the section of the input covered by the kernel.
2. Calculate the element-wise multiplication between the input section and the kernel.
3. Sum the resulting values.
4. Store the sum as one value in the output feature map.

Convolutional Neural Networks (CNNs) use convolution operations to extract spatial features from input data.

Each convolution has a number of parameters:

- Kernel size: Dimensions of the sliding filter.
- Stride: Number of positions the kernel moves between applications.
- Padding: Adds pixels around the input, commonly zeros, to control output dimensions.
- Input channels: Number of channels in the input (e.g. 3 for RGB).
- Output features: Number of filters applied, determining the number of output feature maps.

## Input and output channels

- Each filter operates across all input channels.
- For example, an RGB image has 3 input channels, so a $4 \times 4$ filter has dimensions $4 \times 4 \times 3$.
- Each filter produces one output feature map.
- Therefore, 4 output features means 4 filters and 4 output feature maps.

## Output dimensions

The spatial output dimension can be calculated using:

$$
O = \left\lfloor\frac{I+2P-K}{S}\right\rfloor+1
$$

Where:

- $I$ = input dimension
- $P$ = padding
- $K$ = kernel size
- $S$ = stride

## Hierarchical design in CNNs

CNNs learn hierarchical representations by progressively combining simple features into increasingly complex features.

- Each feature map is an image-like representation containing the activation strengths of a particular filter at different spatial locations.
- Early layers detect simple features (edges, blobs, contrast and curves).
- Intermediate layers detect more complex features (shapes and textures).
- Deeper layers detect object parts (eyes, legs and tyres) and eventually complete objects (faces, dogs and cars).
- Kernel weights are learnt automatically during training through backpropagation and gradient descent.

## Width vs depth

- Width = number of filters in a layer, determining how many different features can be detected.
- Depth = number of layers, determining how many successive levels of feature processing can occur.
- The architecture should be chosen based on the complexity of the dataset and task, aiming to minimise prediction error (task risk).

# Encoders

Encoders typically reduce the spatial dimensions of an input while potentially increasing its feature dimensions.

For example, an image might pass through successive layers that reduce its width and height while increasing the number of feature maps. This produces a more compact representation containing increasingly abstract features.

## Classification

For classification, an encoder extracts features from the input, which are then used by a classification head to predict the probability of each class.

A loss function measures how different the model's predictions are from the correct target. It produces a numerical value representing the model's prediction error.

During training, the model adjusts its parameters to minimise this loss, improving its predictions.

Different tasks use different loss functions:

- Classification: Categorical cross-entropy (CCE) measures how well the model's predicted class probabilities match the correct class.
- Binary classification: Binary cross-entropy (BCE) measures how well the model predicts between two possible classes.
- Regression: L2 loss measures the difference between predicted and actual numerical values, penalising larger errors more heavily.

# Autoencoders

An autoencoder is a neural network that learns a compressed representation of input data and reconstructs the original input from it.

It consists of two main parts:

1. Encoder $E(x)$: Takes the original input $x$ and compresses it into a lower-dimensional representation called the latent representation or latent code, $z$.
2. Decoder $D(z)$: Takes the latent representation $z$ and attempts to reconstruct the original input, producing $D(z)$.

Training aims to make the reconstructed output as similar as possible to the original input.

- A loss function measures the difference between the original input and its reconstruction.
- A common choice is squared L2 loss (reconstruction loss).
- The network learns to minimise this loss.

# Recurrent neural networks

A recurrent neural network (RNN) is a neural network designed to process sequential data by maintaining information from previous timesteps. A timestep is an individual step in a sequence. 

It uses:

- Parameter sharing: RNNs reuse the same weights and biases across multiple timesteps, rather than having separate parameters for each timestep.
- Recurrent connections: Information from previous timesteps is passed forward to influence the processing of the current timestep.
- Unrolling: RNNs can be represented as an unrolled sequence of repeated network operations, making it easier to understand how information flows through the network over time.

# Deep residual networks

Residual connections are shortcut connections that bypass two or three layers (typically containing nonlinear activation functions and batch normalisation) and add the original input directly to the output of those layers.

They:

- Help information and gradients flow through deep networks.
- Make it easier to train very deep neural networks.
- Reduce the vanishing gradient problem, allowing earlier layers to receive stronger gradients during backpropagation.

# Skip connections (U-net)

U-Net is a neural network architecture commonly used for paired image translation tasks, particularly image segmentation.

U-Net uses an encoder-decoder structure with skip connections:

1. Encoder: Compresses the input image, reducing its spatial dimensions while extracting increasingly abstract features.
2. Decoder: Reconstructs an output image, gradually increasing its spatial dimensions.
3. Skip connections: Transfer feature maps directly from encoder layers to corresponding decoder layers.

The skip connections help preserve spatial information and fine details that might otherwise be lost during the encoder's compression.

# Regularisation

Regularisation is the process of adding prior information or constraints about the functions a model can learn to reduce generalisation error.

- Hypothesis space ($H$): The set of all possible functions or mappings that a model could learn to make predictions from its inputs. Regularisation constrains or guides which functions within ($H$) the model is encouraged to learn.

Its purpose is to:

- Prevent overfitting: Discourage models from learning overly complex patterns that fit the training data but do not generalise well to unseen data.
- Control model capacity: Influence the complexity of functions the model can learn.
- Improve generalisation: Encourage the model to learn patterns that are more likely to work well on unseen data.

## Early stopping

Early stopping is a regularisation technique where training is stopped before the model completes all planned epochs (one epoch is one complete pass through the entire training dataset).

- Training loss generally continues to decrease as training progresses.
- Validation loss initially decreases but may start increasing when the model begins to overfit.
- Training is stopped when validation performance stops improving, helping prevent overfitting.

## Annealing

Annealing is the gradual reduction of the learning rate towards zero during training.

- As the learning rate decreases, the model makes progressively smaller parameter updates.
- This has a similar effect to early stopping because it gradually reduces how much the model can change.
- However, it allows training to continue for longer, giving the model more opportunity to learn from the data.

## Data augmentation

Data augmentation is a regularisation technique that artificially increases the size and diversity of the training dataset by applying transformations to existing training samples.

Examples of transformations include:

- Rotations
- Zooming
- Flipping
- Blurring

Data augmentation:

- Helps the model generalise to variations it may encounter in unseen data.
- Reduces overfitting by exposing the model to a wider range of training examples.
- Is particularly useful when the original training dataset is limited.

However:

- Too much augmentation (e.g. excessive zooming) can make training examples unrealistic, resulting in poor fitting.
- Extreme transformations may change the class label. For example, rotating handwritten MNIST digits by $180^\circ$ can change their identity (e.g. a 6 may become a 9).
- Augmentations should reflect transformations that are reasonably expected in testing.

## Dropout

Dropout is a regularisation technique where each hidden unit is randomly set to zero with a given probability during training (e.g. 0.2):

- Prevents the network from relying too heavily on any individual hidden unit.
- Encourages the network to distribute information across multiple units.
- Reduces overfitting and improves generalisation.

## Tikhonov regularisation (L2 regularisation / weight decay)

Tikhonov regularisation adds a penalty to the loss function based on the squared magnitude of the model's weights.

- Penalises large weights, encouraging the model to use smaller weights.
- Discourages overly complex models and helps prevent overfitting.
- The strength of the penalty is controlled by the regularisation parameter $\lambda$.

## Ensembles

An ensemble is a machine learning technique where multiple different models are trained and their predictions are combined at test time. 

There are two main ways in which predictions are combined:

- Averaging: Combines the predictions of multiple models by calculating their average (commonly used for regression).
- Max voting (majority voting): Selects the class predicted by the majority of models (commonly used for classification).

Its purpose it to:

- Combine the strengths of multiple models.
- Improve prediction accuracy and generalisation compared with using a single model.
- Capture different patterns or solutions that individual models might miss.
