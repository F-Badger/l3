## Tensors

A tensor is a multidimensional array of numerical values.

- A scalar has 0 dimensions (no axes), a vector has 1 and a matrix has 2.
  - a scalar is a single number, a vector is a sequence of numbers, a matrix is a table of numbers arranged in rows and columns
- The shape gives the size of each axis.
- An image batch typically has four axes: batch size, channels, height and width.

A batch groups examples so they can be processed together using the same model.

```python
import torch

x = torch.zeros(10) # create a vector (1D tensor) of length 10, filled with zeros

print(x)
x.size()
```

## Broadcasting
Broadcasting is a PyTorch feature that allows tensors with different shapes to be used together in element-wise operations, provided their shapes are compatible. PyTorch automatically expands the smaller tensor logically, without necessarily copying its data.

```python
x+2 # operations can be performed on all elements
```

```python
id = torch.eye(10) # creates an identity matrix (2D tensor) of the specified size
id
```

```python
x+3 * id
```

```python
torch.rand(10,5) # creates a tensor of size (10,5) filled with random values between 0 and 1
```

## Adding and removing dimensions

`unsqueeze()` adds a dimension of size 1 to a tensor's shape without changing its data. This is useful when you need a tensor to have a particular shape for operations such as broadcasting. The argument specifies where the new dimension is inserted.

```python
x.size()
```

```python
x = x.unsqueeze(0) # add a dimension at axis 0
x.size()
```

```python
x = x.unsqueeze(2) # add a dimension at axis 1
x.size()
```

`squeeze()` can be used to remove dimensions of size 1 from a tensor's shape without changing its data. It is essentially the opposite of `unsqueeze()`. With no argument, all singleton dimensions (dimensions of size 1) are removed. The argument can be used to specify which dimension to remove. If the specified dimension is not size 1, `squeeze(dim)` leaves the shape unchanged.

```python
x = x.squeeze(0)
x.size()
```

```python
x = x.squeeze()
x.size()
```

## Repeating values

`repeat()` repeats a tensor's data along specified dimensions, creating a larger tensor.

The arguments specify how many times to repeat the data along each axis.

```python
y = torch.tensor([[1, 2],
                  [3, 4]])
y = y.repeat(2, 3) # repeat twice along first axis, three time along second axis.
y
```

## Indexing

Indexing lets you access specific elements, rows, columns or slices of a tensor using square brackets `[]`. 

- Indexing starts at 0, not 1
- Negative indices count from the end: -1 is the last element
- The colon `:` means select all elements along that dimension
  - For example, `x[:, 1]` selects every row but only column 1
- Slicing selects a range of elements using `start:stop:step`

```python
y
```

```python
y.size()
```

```python
y[0]
```

```python
y[0].size()
```

```python
y[:,0] # select all rows, 0th (first) element
```

```python
y[:3] # select first three rows
```

```python
y[3:] # select all rows after row 3
```

## Changing tensor shape
`view()` changes the shape of a tensor without changing its data or the order of its elements. The new shape must contain the same total number of elements as the original tensor. The product of the dimensions must equal the original number of elements.

```python
a = torch.tensor([1, 2, 3, 4, 5, 6]) # 1D tensor with 6 elements
print(a.shape)

a = a.view(2, 3) # Rearrange into two rows, three columns
print(a)
print(a.shape)
```

```python
b = torch.tensor([[[1],[2],[3]],
                   [[4],[5],[6]]])
b.size()
```

```python
b.view(-1) # -1 can be used for one dimension and PyTorch will calculate its size automatically
```

```python
b.view(3,-1)
```

## Models and loss

A model $f_\theta$ maps an input $x$ to a prediction $f_\theta(x)$. Its parameters $\theta$ are learned from training data.

For $n$ training pairs $(x_i, y_i)$, the mean squared error (MSE) measures the average squared difference between the predictions and the target values:

$$
\mathcal{L}(\theta)
=
\frac{1}{n}
\sum_{i=1}^{n}
\left(f_\theta(x_i)-y_i\right)^2
$$

- $x_i$: the $i$-th input.
- $y_i$: the corresponding target value.
- $f_\theta(x_i)$: the model's prediction.
- $\mathcal{L}(\theta)$: the loss, measuring prediction error.

Training adjusts the parameters $\theta$ to minimise the loss on the training examples. Lower MSE indicates more accurate predictions on average, with larger errors penalised more heavily because the differences are squared.

## Training gradients and parameter updates

Training adjusts a model's parameters $\theta$ to minimise its loss using three steps:

1. Forward pass: compute predictions from the inputs, then calculate a scalar loss measuring the prediction error.
2. Backpropagation: use the chain rule to calculate the gradient of the loss with respect to each parameter. The gradient indicates how changing each parameter affects the loss.
3. Parameter update: adjust the parameters in the direction of the negative gradient to reduce the loss.

For a batch with mean loss $\mathcal{L}_B(\theta)$, gradient descent updates the parameters using

$$
\theta \leftarrow \theta - \eta \nabla_\theta \mathcal{L}_B(\theta)
$$

- $\theta$: model parameters.
- $\nabla_\theta \mathcal{L}_B(\theta)$: gradient of the batch loss with respect to the parameters.
- $\eta$: **learning rate**, which controls the step size.

A larger learning rate makes larger updates, while a smaller learning rate makes smaller updates. If the learning rate is too large, training may become unstable; if it is too small, training may be slow.

## Tracking operations

In PyTorch, `requires_grad=True` tells PyTorch to track operations on a tensor so it can calculate gradients for it during backpropagation.

```python
c = torch.zeros(16,16)
c.requires_grad = True # Enable gradient tracking

d = c+2 # PyTorch records this operation in the computational graph.

# Inspect the function that produced d.
# AddBackward0 represents the addition operation in the graph.
d.grad_fn
```

```python
d = d*3 # PyTorch records this multiplication in the computational graph.

# Inspect the function that produced the new d.
# MulBackward0 represents the multiplication operation in the graph.
d.grad_fn
```

```python
# Inspect the previous operation in the computational graph.
# next_functions[0][0] points to the function that produced the
# input to the multiplication: the addition operation.
d.grad_fn.next_functions[0][0]
```

```python
# Calculate the mean of all elements in d, producing a scalar tensor.
out = d.mean()
out
```

```python
# Check the gradient of c before backpropagation.
# It is None because the gradient has not yet been calculated.
c.grad
```

```python
# Perform backpropagation to calculate the gradient of out with respect to c.
out.backward()

# Inspect the gradient of out with respect to each element of c.
c.grad
```
